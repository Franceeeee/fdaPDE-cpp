```python
import os
import numpy as np
import pandas as pd
```

```python
# Crea la directory fine_mesh se non esiste
output_dir = "coarsest_mesh"
os.makedirs(output_dir, exist_ok=True)
```

```python
# Funzione per generare una mesh su un quadrato unitario con n x n punti
def generate_mesh(n):
    x = np.linspace(0, 1, n)
    y = np.linspace(0, 1, n)
    X, Y = np.meshgrid(x, y)
    points = np.vstack([X.ravel(), Y.ravel()]).T
    return points, X, Y

# Funzione per generare gli elementi (triangoli)
def generate_elements(n):
    elements = []
    for i in range(n - 1):
        for j in range(n - 1):
            p1 = i * n + j + 1
            p2 = p1 + 1
            p3 = p1 + n
            p4 = p3 + 1
            elements.append([p1, p2, p3])  # Triangolo inferiore
            elements.append([p2, p4, p3])  # Triangolo superiore
    return elements

# Funzione per identificare i vicini degli elementi
def generate_neighbors(elements):
    neighbors = []
    for i in range(len(elements)):
        neighbors.append([-1, -1, -1])  # Placeholder per i vicini
    return neighbors

# Funzione per generare gli spigoli della mesh
def generate_edges(elements):
    edges = set()
    for element in elements:
        edges.add((min(element[0], element[1]), max(element[0], element[1])))
        edges.add((min(element[1], element[2]), max(element[1], element[2])))
        edges.add((min(element[2], element[0]), max(element[2], element[0])))
    return list(edges)

# Funzione per identificare i punti di bordo
def generate_boundary(points, n):
    boundary = []
    for i, point in enumerate(points):
        if point[0] == 0 or point[0] == 1 or point[1] == 0 or point[1] == 1:
            boundary.append(1)
        else:
            boundary.append(0)
    return boundary
```

```python
# Numero di punti per lato (modificabile)
n = 10  # Un quadrato 10x10

# Genera la mesh e gli elementi
points, X, Y = generate_mesh(n)
elements = generate_elements(n)
neighbors = generate_neighbors(elements)
edges = generate_edges(elements)
boundary = generate_boundary(points, n)

# Salva points.csv nel formato corretto
points_df = pd.DataFrame(points, columns=["V1", "V2"])
points_df.index += 1
points_df.to_csv(os.path.join(output_dir, "points.csv"), index_label="", float_format="%.16f", quoting=2)

# Salva elements.csv
elements_df = pd.DataFrame(elements, columns=["V1", "V2", "V3"])
elements_df.index += 1
elements_df.to_csv(os.path.join(output_dir, "elements.csv"), index_label="",quoting=2)

# Salva neigh.csv (con valori -1 come placeholder)
neighbors_df = pd.DataFrame(neighbors, columns=["V1", "V2", "V3"])
neighbors_df.index += 1
neighbors_df.to_csv(os.path.join(output_dir, "neigh.csv"), index_label="",quoting=2)

# Salva edges.csv
edges_df = pd.DataFrame(edges, columns=["V1", "V2"])
edges_df.index += 1
edges_df.to_csv(os.path.join(output_dir, "edges.csv"), index_label="",quoting=2)

# Salva boundary.csv
boundary_df = pd.DataFrame(boundary, columns=["V1"])
boundary_df.index += 1
boundary_df.to_csv(os.path.join(output_dir, "boundary.csv"), index_label="",quoting=2)

print(f"Files generated in the '{output_dir}' directory: points.csv, elements.csv, neigh.csv, edges.csv, boundary.csv")
```
